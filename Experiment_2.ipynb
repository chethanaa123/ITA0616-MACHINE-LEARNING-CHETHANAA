```python

import pandas as pd
from itertools import product

# Create the training dataset
data = {
    'Size': ['Big', 'Small', 'Small', 'Big', 'Small'],
    'Color': ['Red', 'Red', 'Red', 'Blue', 'Blue'],
    'Shape': ['Circle', 'Triangle', 'Circle', 'Circle', 'Circle'],
    'Class': ['No', 'No', 'Yes', 'No', 'Yes']
}

pd.DataFrame(data).to_csv('training.csv', index=False)

# Read the CSV file
df = pd.read_csv('training.csv')
attributes = ['Size', 'Color', 'Shape']
domains = [df[a].unique().tolist() + ['?'] for a in attributes]

# Generate all possible hypotheses
consistent = []

for h in product(*domains):
    valid = True

    for _, row in df.iterrows():
        match = all(
            h[i] == '?' or h[i] == row[a]
            for i, a in enumerate(attributes)
        )

        if (row['Class'] == 'Yes' and not match) or \
           (row['Class'] == 'No' and match):
            valid = False
            break

    if valid:
        consistent.append(list(h))

# Display the result
print("Training Dataset:")
print(df.to_string(index=False))

print("\nConsistent Hypotheses:")
for h in consistent:
    print(h)

print("\nTotal Consistent Hypotheses:", len(consistent))
```
